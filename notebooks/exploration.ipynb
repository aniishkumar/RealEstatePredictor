# California Housing exploration
Run `python -m ml.train` first. This notebook reads the exact processed copy produced by training.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/california_housing.csv')
print('Shape:', df.shape)
display(df.dtypes)
display(df.isna().sum())
print('Duplicates:', df.duplicated().sum())
display(df.describe().T)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['median_house_value'].hist(ax=axes[0], bins=40)
axes[0].set(title='Target distribution', xlabel='Median house value ($100k)')
axes[1].scatter(df['median_income'], df['median_house_value'], s=3, alpha=.2)
axes[1].set(title='Income vs target', xlabel='Median income', ylabel='Median house value ($100k)')
plt.tight_layout()
```

## Questions answered
- Are there missing values or duplicate rows?
- How wide and skewed is the target distribution?
- Is median income associated with the target? Correlation is not causation.
- Are implausibly large household averages present and worth handling through robust preprocessing?
